```python
# convert whole size values in bytes
def convert_size(size):
    if isinstance(size , str):
        if 'k' in size:
            return float(size.replace('k', '')) * 1024
        elif 'M' in size:
            return float(size.replace('M', '')) * 1024 * 1024
        elif 'Varies with device' in size:
            return np.nan
    return size
```

```python
df['Size'] = df['Size'].apply(lambda x: convert_size(x) if isinstance(x, str) else x)
```

```python
# change column name
df.rename(columns={'Size':'size in bytes'}, inplace=True)
```

```python
df.head()
```

```python
df['size _in_Mb'] = df['size in bytes'].apply(lambda x:x /(1024 * 1024))
```

```python
df['size-in-kb'] = df['size in bytes'].apply(lambda x: x/1024)
```

# let's take care of install values

```python
df['Installs'].unique()
```

```python
df['Installs'].isnull().sum()
```

```python
# remove the '+' and ',' from the 'Installs' column
df['Installs'] = df['Installs'].str.replace('+', '').str.replace(',', '')
```

```python
df.head()
```

```python
# convert its value into integer
df['Installs'] = pd.to_numeric(df['Installs'], errors='coerce')
```

```python
df['Installs'].unique()
```

# price column

```python
df['Price'].value_counts()
```

```python
# how many values have $ sign
df['Price'].str.contains('$').sum()
# remove the $ sign and convert to float
df['Price'] = df['Price'].str.replace('$', '').astype(float)
```
